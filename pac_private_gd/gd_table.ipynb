```python
import pickle
import numpy as np

def get_all_accs(dataset, e0, budgets):
    accs = {}
    for ind in range(len(budgets)):
        tmp = pickle.load(open(f'results/{dataset}_data_budget={int(budgets[ind])}_e0={e0}.pkl', 'rb'))
        l1 = [k[0] for k in tmp[False]]
        l2 = [k[0] for k in tmp[True]]
        accs[int(budgets[ind])] = (l1, l2)
        
    return accs

names = ['Credit', 'MNIST (Easy)', 'MNIST (Hard))']
for data_ind, data in enumerate(['credit', 'mnist_0_vs_7', 'mnist_7_vs_9']):

    # print(f'### {names[data_ind]} ###')
    mi_range = [2**(x) for x in range(-10, 0, 2)]

    labels = [-np.log2(x) for x in mi_range]

    budget_range = [1/(mi) for mi in mi_range]
    budget_range.reverse()
    formatted = {}
    for e0 in ['exact', 0.01, 0.1, 1.0]:
        # for aware in [False, True]:
        row_res = []
        if e0 != 'exact':
            row_res.append('')
            # row_res.append(f'{e0}' if not aware else '')
            row_res.append(f'{e0}')
        else:
            # row_res.append(names[data_ind])
            row_res.append(r'\multirow{4}{*}{' + names[data_ind] + r'}')
            row_res.append(r'Oracle')

        orig_accs = get_all_accs(data, e0, budget_range)
        for budget in budget_range:
            obl_accs = orig_accs[budget][0]
            obl_mean = np.average(obl_accs)
            aware_accs = orig_accs[budget][1]
            aware_mean = np.average(aware_accs)
            # fin = '\\tup' + "{" + f'{round(obl_mean, 2)}' + "}" + "{" + f'{round(aware_mean, 2)}' + "}"
#             +\{{round(aware_mean, 2)}\}'
            # fin = f''
            # (round(obl_mean, 2), round(aware_mean, 2))
            fin = f'({round(obl_mean, 2)}, {round(aware_mean, 2)})'
            row_res.append(str(fin))

        
        print(' & '.join(row_res) + "\\\\ ")
    # print('-------')
    #     print(formatted)
```
